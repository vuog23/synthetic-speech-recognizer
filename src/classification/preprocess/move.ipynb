```python
import os
import shutil
import pandas as pd
```

```python
asv_root = r"D:\Project\Synthetic Speech Recognizer\datasets\ASVspoof_2019_LA\processed"
libri_root = r"D:\Project\Synthetic Speech Recognizer\datasets\LibriSpeech_100\processed"
save_root = r"D:\Project\Synthetic Speech Recognizer\datasets\ASVLibri"
```

```python
asv_splits = os.listdir(asv_root)
libri_split = os.listdir(libri_root)
```

```python
for split in asv_splits:
    split_path = os.path.join(asv_root, split)
    for cat in os.listdir(split_path):
        cat_path = os.path.join(split_path, cat)
        for wav in os.listdir(cat_path):
            wav_path = os.path.join(cat_path, wav)
            save_path = os.path.join(save_root, split, cat)
            shutil.copy(wav_path, save_path)
```

```python
for split in libri_split:
    split_path = os.path.join(libri_root, split)
    save_path = os.path.join(save_root, split, "real")
    os.makedirs(save_path, exist_ok=True)

    for wav in os.listdir(split_path):
        wav_path = os.path.join(split_path, wav)
        if os.path.isfile(wav_path):
            shutil.copy(wav_path, save_path)
```
