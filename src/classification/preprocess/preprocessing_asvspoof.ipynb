```python
import pandas as pd
import io
import soundfile as sf
import os
```

```python
root_path = r"D:\Project\Synthetic Speech Recognizer\datasets\ASVspoof_2019_LA\processed"
train_df = pd.read_parquet(r"D:\Project\Synthetic Speech Recognizer\datasets\ASVspoof_2019_LA\raw\train-00000-of-00001.parquet")
test_df = pd.read_parquet(r"D:\Project\Synthetic Speech Recognizer\datasets\ASVspoof_2019_LA\raw\test-00000-of-00001.parquet")
val_df = pd.read_parquet(r"D:\Project\Synthetic Speech Recognizer\datasets\ASVspoof_2019_LA\raw\validation-00000-of-00001.parquet")
```

```python
def byte2wave(df, index, split, root_path):
    audio_bytes = df.iloc[index]["audio"]["bytes"]

    waveform, sample_rate = sf.read(
        io.BytesIO(audio_bytes)
    )

    if df.iloc[index]["key"] == 0:
        label = "real"
    else:
        label = "fake"

    file_name = df.iloc[index]["audio_file_name"]

    save_path = os.path.join(root_path, split, label, file_name + ".wav")

    sf.write(
        save_path,
        waveform,
        sample_rate
    )
```

```python
for i in range(len(train_df)):
    byte2wave(train_df, i, "train", root_path)
```

```python
for i in range(len(test_df)):
    byte2wave(test_df, i, "test", root_path)
```

```python
for i in range(len(val_df)):
    byte2wave(val_df, i, "val", root_path)
```
