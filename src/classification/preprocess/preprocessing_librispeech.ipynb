```python
import pandas as pd
import io
import soundfile as sf
import os
```

```python
test_path = r"D:\Project\Synthetic Speech Recognizer\datasets\LibriSpeech_100\raw\test-0000.parquet"
val_path = r"D:\Project\Synthetic Speech Recognizer\datasets\LibriSpeech_100\raw\val-0000.parquet"
test_df = pd.read_parquet(test_path)
val_df = pd.read_parquet(val_path)
```

```python
save_test_path = r"D:\Project\Synthetic Speech Recognizer\datasets\LibriSpeech_100\processed\test"
save_val_path = r"D:\Project\Synthetic Speech Recognizer\datasets\LibriSpeech_100\processed\val"
```

```python
def byte2wave(df, index, save_path):
    audio_bytes = df.iloc[index]["audio"]["bytes"]

    waveform, sample_rate = sf.read(
        io.BytesIO(audio_bytes)
    )

    file_name = "libri_" + df.iloc[index]["id"].replace("-", "_")

    save_path = os.path.join(save_path, file_name + ".wav")

    sf.write(
        save_path,
        waveform,
        sample_rate
    )
```

```python
for i in range(len(test_df)):
    byte2wave(test_df, i, save_test_path)
```

```python
for i in range(len(val_df)):
    byte2wave(val_df, i, save_val_path)
```
